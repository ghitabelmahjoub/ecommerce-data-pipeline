```python
import pandas as pd

# Load the raw dataset
df = pd.read_excel("data/raw/Online Retail.xlsx")

print("Dataset loaded successfully!") 
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
df.head()
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing values per column:")
print(missing_values)
```

```python
# Check for duplicate rows
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
# Check quantity statistics
print("Minimum quantity:", df["Quantity"].min())
print("Maximum quantity:", df["Quantity"].max())
print("Negative quantity rows:", (df["Quantity"] < 0).sum())
print("Zero quantity rows:", (df["Quantity"] == 0).sum())
```

```python
# Check cancelled invoices
cancelled_invoices = df["InvoiceNo"].astype(str).str.startswith("C")

print("Cancelled invoice rows:", cancelled_invoices.sum())
print("Negative quantity rows:", (df["Quantity"] < 0).sum())

print("\nExamples of cancelled invoices:")
print(df.loc[cancelled_invoices].head())
```

```python
# Check unit price values

print("Minimum UnitPrice:", df["UnitPrice"].min())
print("Maximum UnitPrice:", df["UnitPrice"].max())
print("Zero UnitPrice rows:", (df["UnitPrice"] == 0).sum())
print("Negative UnitPrice rows:", (df["UnitPrice"] < 0).sum())
```

```python
# Inspect rows with negative UnitPrice
negative_price = df[df["UnitPrice"] < 0]

print(negative_price)
```

```python
# Inspect rows with zero UnitPrice
zero_price = df[df["UnitPrice"] == 0]

print("Number of zero-price rows:", len(zero_price))
print("\nSample of zero-price rows:")
display(zero_price.head(20))
```

```python
# Inspect zero-price transactions
zero_price = df[df["UnitPrice"] == 0]

print("Number of zero-price rows:", len(zero_price))

print("\nDescriptions:")
print(zero_price["Description"].value_counts().head(20))

print("\nSample rows:")
display(zero_price.head(20))
```

```python
# Analyze quantities of zero-price transactions

print("Quantity distribution for zero-price rows:")
print(zero_price["Quantity"].value_counts().head(20))

print("\nPositive quantity:", (zero_price["Quantity"] > 0).sum())
print("Negative quantity:", (zero_price["Quantity"] < 0).sum())
print("Zero quantity:", (zero_price["Quantity"] == 0).sum())
```

```python
# Inspect zero-price rows with positive quantities

zero_price_positive = zero_price[zero_price["Quantity"] > 0]

print("Zero-price + positive quantity:", len(zero_price_positive))

print("\nDescriptions:")
print(zero_price_positive["Description"].value_counts().head(20))

print("\nSample:")
display(zero_price_positive.head(20))
```

```python
# Inspect zero-price rows with negative quantities

zero_price_negative = zero_price[zero_price["Quantity"] < 0]

print("Zero-price + negative quantity:", len(zero_price_negative))

print("\nDescriptions:")
print(zero_price_negative["Description"].value_counts().head(20))

print("\nSample:")
display(zero_price_negative.head(20))
```

```python
# ==========================================
# STEP 2: CLEAN THE DATA
# ==========================================

# Make a copy of the original dataset
clean_df = df.copy()

print("Original rows:", len(clean_df))

# 1. Remove duplicate rows
clean_df = clean_df.drop_duplicates()

print("After removing duplicates:", len(clean_df))

# 2. Remove rows with missing Description
clean_df = clean_df.dropna(subset=["Description"])

print("After removing missing descriptions:", len(clean_df))

# 3. Remove cancelled invoices
clean_df = clean_df[
    ~clean_df["InvoiceNo"].astype(str).str.startswith("C")
]

print("After removing cancelled invoices:", len(clean_df))

# 4. Remove negative quantities
clean_df = clean_df[clean_df["Quantity"] > 0]

print("After removing negative/zero quantities:", len(clean_df))

# 5. Remove zero or negative prices
clean_df = clean_df[clean_df["UnitPrice"] > 0]

print("After removing invalid prices:", len(clean_df))

print("\nFinal cleaned shape:", clean_df.shape)
```

```python
# ==========================================
# STEP 3: VALIDATE CLEANED DATA
# ==========================================

print("Missing values:")
print(clean_df.isnull().sum())

print("\nDuplicate rows:")
print(clean_df.duplicated().sum())

print("\nNegative quantities:")
print((clean_df["Quantity"] < 0).sum())

print("\nZero quantities:")
print((clean_df["Quantity"] == 0).sum())

print("\nInvalid prices (<= 0):")
print((clean_df["UnitPrice"] <= 0).sum())

print("\nCancelled invoices:")
print(
    clean_df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .sum()
)
```

```python
import os

# Make sure the processed folder exists
os.makedirs("data/processed", exist_ok=True)

# Save cleaned dataset
output_path = "data/processed/online_retail_cleaned.csv"

clean_df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("Location:", output_path)

# Check file size
file_size_mb = os.path.getsize(output_path) / (1024 * 1024)
print(f"File size: {file_size_mb:.2f} MB")
```
